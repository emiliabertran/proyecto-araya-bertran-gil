```python
file_path = 'bertran.database2.limpia.csv'
df = pd.read_csv(file_path, sep=";")
source = df.groupby('Comuna de crianza').size().reset_index(name='Cantidad')
base = alt.Chart(source).encode(
    x=alt.X('Cantidad:Q').stack('zero'),
    y=alt.Y('Comuna de crianza:N', title='Comuna').sort('-x'),
    text=alt.Text('Cantidad:Q', format='.0f')
)
bars = base.mark_bar().encode(
    color=alt.Color('Cantidad:Q',
                    scale=alt.Scale(range=['#f472b6','#3b82f6']),
                    legend=None)
)
text = base.mark_text(
    align='right',
    dx=-3,
    color='white'
)
grafico_final = (bars + text).properties(
    title='Ganadoras por Comuna',
    width=500,
    height=400
)
grafico_final.save('visualizacion_final.html')
grafico_final
```

```python
file_path = 'bertran.database2.limpia.csv'
df = pd.read_csv(file_path, sep=";")
source = df.groupby('IDH comunal').size().reset_index(name='Cantidad')
orden_idh = ['Muy Alto', 'Alto', 'Medio-Alto', 'Medio', 'Medio-Bajo', 'Bajo']
base = alt.Chart(source).encode(
    x=alt.X('Cantidad:Q', title='Número de Ganadoras').stack('zero'),
    y=alt.Y('IDH comunal:N', title='Nivel de IDH', sort=orden_idh),
    text=alt.Text('Cantidad:Q', format='.0f')
)
bars = base.mark_bar().encode(
    color=alt.Color('Cantidad:Q',
                    scale=alt.Scale(range=['#f472b6','#3b82f6']),
                    legend=None)
)
text = base.mark_text(
    align='right',
    dx=-3,
    color='white'
)
grafico_final = (bars + text).properties(
    title='Distribución de Ganadoras según IDH de la Comuna',
    width=500,
    height=400
)
grafico_final.save('visualizacion_final_idh.html')
grafico_final
```
