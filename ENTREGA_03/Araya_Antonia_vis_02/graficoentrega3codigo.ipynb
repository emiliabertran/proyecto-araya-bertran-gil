```python
import pandas as pd
import altair as alt
from google.colab import files


df = pd.read_csv('experienciamiss.csv')


df_final = df.copy()
df_final.columns = ['Nombre', 'Anio', 'Certamen', 'Categoria', 'Detalle_M', 'Cat_TV', 'Detalle_TV', 'Vinculos']
df_final['Decada'] = (df_final['Anio'] // 10) * 10
df_final['Decada'] = df_final['Decada'].astype(str) + "s"


grafico = alt.Chart(df_final).mark_bar(size=60).encode(
    x=alt.X('Decada:N', title='Décadas', sort=['1990s', '2000s', '2010s', '2020s']),
    y=alt.Y('count():Q', title='Cantidad de Ganadoras'),
    color=alt.Color('Categoria:N',
                    title='Nivel de Experiencia',
                    scale=alt.Scale(domain=['Profesional', 'Amateur', 'Sin experiencia'],
                                    range=['#1f77b4', '#ff7f0e', '#2ca02c'])),
    tooltip=['count()', 'Decada', 'Categoria']
).properties(
    title='Evolución del Capital Previo por Década (Miss Universo Chile)',
    width=500, height=400
)


grafico.display()
grafico.save('vis_02.html')
df_final.to_csv('experienciamiss_final.csv', index=False)

print("---")
print("⬇️ Se están descargando tus archivos...")
files.download('vis_02.html')
files.download('experienciamiss_final.csv')
```
