```python
import pandas as pd
import altair as alt
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
df = pd.read_excel("Data de base limpia.xlsx")

df.head()
```

```python
df.columns
```

```python
df.info()

df.isnull().sum()
```

```python
grafico = alt.Chart(df).mark_bar().encode(
    x=alt.X(
        'count():Q',
        title='Cantidad de participantes'
    ),

    y=alt.Y(
        'Trayectoria post certamen:N', # Corrected column name
        sort='-x',
        title='Trayectoria posterior'
    ),

    tooltip=[
        'Trayectoria post certamen', # Corrected column name
        'count()'
    ]
).properties(
    title='Trayectoria posterior de participantes de Miss Universo Chile',
    width=700,
    height=400
)

grafico
```

```python
grafico.save('vis_01.html')
```
