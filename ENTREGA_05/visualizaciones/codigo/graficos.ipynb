```python
import pandas as pd
import altair as alt

df = pd.read_csv('base_miss_chile_unificada.csv')

grafico_geografico = alt.Chart(df).mark_bar(color='#122347').encode( # <-- Azul Marino
    x=alt.X('count():Q', title='Cantidad de Ganadoras'),
    y=alt.Y('Región:N', title='Región de Origen', sort='-x'),
    tooltip=[
        alt.Tooltip('Región:N', title='Región'),
        alt.Tooltip('count():Q', title='Cantidad de Ganadoras')
    ]
).properties(
    title='El Centralismo de la Corona: Ganadoras por Región de Origen (1952-2025)',
    width=500, height=350
).configure_view(
    fill='#F9F9FB'
)

grafico_geografico.display()
grafico_geografico.save('grafico_geografico_regiones.html')
```

```python
import pandas as pd
import altair as alt

# 1. Cargamos los datos y configuramos la década
df = pd.read_csv('base_miss_chile_unificada.csv')
df['Decada'] = (df['Año'] // 10) * 10
df['Decada'] = df['Decada'].astype(str) + "s"

columna_destino = 'Medios-Postcertamen'
df[columna_destino] = df[columna_destino].astype(str)

# 2. Creamos el gráfico con la paleta sobria (Azul Marino y Morado Profundo)
grafico_destino_tiempo = alt.Chart(df).mark_bar().encode(
    x=alt.X('Decada:N', title='Décadas', sort='ascending'),
    y=alt.Y('count():Q', title='Cantidad de Mujeres'),
    color=alt.Color(f'{columna_destino}:N',
                    title='¿Ingresó a los Medios?',
                    scale=alt.Scale(range=['#122347', '#311B92'])), # <-- Azul Marino para "No", Morado Profundo para "Sí"
    column=alt.Column('Certamen:N', title='Evolución del Destino por Tipo de Certamen'),
    tooltip=[
        alt.Tooltip('count()', title='Cantidad'),
        alt.Tooltip('Decada', title='Década'),
        alt.Tooltip('Certamen', title='Certamen'),
        alt.Tooltip(f'{columna_destino}:N', title='Destino')
    ]
).properties(
    width=180, height=300
)

# Mostrar en pantalla y guardar el archivo definitivo
grafico_destino_tiempo.display()
grafico_destino_tiempo.save('grafico_destino_postcertamen.html')
```

```python
import pandas as pd
import altair as alt

# 1. Cargamos los datos y configuramos la década
df = pd.read_csv('base_miss_chile_unificada.csv')
df['Decada'] = (df['Año'] // 10) * 10
df['Decada'] = df['Decada'].astype(str) + "s"

columna_estudios = 'Nivel-Estudios'
df[columna_estudios] = df[columna_estudios].astype(str)

# 2. Creamos el gráfico con la paleta sobria de gala
grafico_estudios = alt.Chart(df).mark_bar(size=25).encode(
    x=alt.X('Decada:N', title='Décadas', sort='ascending'),
    y=alt.Y('count():Q', title='Cantidad de Ganadoras'),
    # Asignamos la nueva combinación: Morado, Azul Marino, Dorado y Gris
    color=alt.Color(f'{columna_estudios}:N',
                    title='Nivel Educacional',
                    scale=alt.Scale(range=['#311B92', '#122347', '#C5A059', '#7A8B99'])),
    tooltip=[
        alt.Tooltip('count()', title='Cantidad'),
        alt.Tooltip('Decada', title='Década'),
        alt.Tooltip(f'{columna_estudios}:N', title='Estudios')
    ]
).properties(
    title='Evolución del Nivel de Estudios de las Ganadoras por Década',
    width=600, height=400
)

# Mostrar en pantalla y guardar el archivo interactivo
grafico_estudios.display()
grafico_estudios.save('grafico_nivel_estudios.html')
```

```python
import pandas as pd
import altair as alt
from google.colab import files

df = pd.read_csv('base_miss_chile_unificada.csv')
df['Decada'] = (df['Año'] // 10) * 10
df['Decada'] = df['Decada'].astype(str) + "s"

# Paleta de alto contraste sobre fondo oscuro: Dorado, Azul Acero, Blanco Invierno, Gris
colores_estudios = ['#C9A86A', '#35568A', '#FAF9F6', '#7A8B99']

grafico_estudios = alt.Chart(df).mark_bar().encode(
    x=alt.X('Decada:N', title='Décadas', sort='ascending', axis=alt.Axis(labelAngle=0)),
    y=alt.Y('count():Q', title='Cantidad de Ganadoras'),
    color=alt.Color('Nivel Educacional:N',
                    title='Nivel Educacional',
                    scale=alt.Scale(range=colores_estudios)),
    tooltip=[alt.Tooltip('Decada:N'), alt.Tooltip('count()'), alt.Tooltip('Nivel Educacional:N')]
).properties(
    title='Evolución del Nivel de Estudios de las Ganadoras por Década',
    width=550, height=380
).configure(
    background='transparent',
    view=alt.ViewConfig(stroke='transparent')
)

grafico_estudios.save('grafico_nivel_estudios.html')
files.download('grafico_nivel_estudios.html')
```
